# 저장 모델 평가

공통 평가 진입점은 `evaluation/evaluation.py`입니다. `evaluation/eval.yaml`에서 ML, DL, fusion 대상, artifact 경로, split, threshold override, 결과 위치를 선택합니다. test 결과를 보며 모델이나 threshold를 조정하지 마세요.

DL 가중치는 `evaluation.model_path`에 run 폴더 또는 `best_model.pt`를 지정합니다. 구조·전처리 복원용 metadata와 config 파일은 같은 폴더에 보관합니다. 평가 시 학습을 수행하지 않습니다.

```python
import json
import sys
from pathlib import Path

import pandas as pd

repo = Path.cwd().resolve()
while not (repo / "Members/sang/model/evaluation/eval.yaml").is_file():
    if repo.parent == repo:
        raise RuntimeError("저장소 루트에서 노트북을 실행하세요")
    repo = repo.parent
eval_config = repo / "Members/sang/model/evaluation/eval.yaml"
if str(repo) not in sys.path:
    sys.path.insert(0, str(repo))
from Members.sang.model.evaluation.evaluation import evaluate_saved_model  # noqa: E402
```

```python
evaluation_run = evaluate_saved_model(eval_config)
print(evaluation_run)
print(json.loads((evaluation_run / "metrics.json").read_text()))
filename = (
    "test_predictions.csv"
    if (evaluation_run / "test_predictions.csv").exists()
    else "predictions.csv"
)
display(pd.read_csv(evaluation_run / filename).head())
```
