```python
import sys
from io import StringIO

import matplotlib.pyplot as plt
import pandas as pd

print(sys.version)
print(sys.executable)
frame = pd.read_csv(StringIO("subject_id,value\nsynthetic_1,10\nsynthetic_2,\nsynthetic_3,30\n"))
frame
```

```python
frame.isna().sum()
```
